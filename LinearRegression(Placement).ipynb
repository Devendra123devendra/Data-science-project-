# import library

```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np
```

# import dataset

```python
df = pd.read_csv('placement (1).csv')
df.head()
```

# data plot in graph

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel('pakage(in lpa)')
```

# Define the x and y as a input and output column

```python
x = df.iloc[:,0:1]
y = df.iloc[:,-1]
```

```python
x
```

```python
y
```

# train test split 

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state = 2)
```

# import linear regression 

```python
from sklearn.linear_model import LinearRegression
```

# define lr

```python
lr = LinearRegression()
```

# use fit for trine the model 

```python
lr.fit(x_train,y_train)
```

```python
x_test
```

```python
y_test
```

# predict value

```python
lr.predict(x_test.iloc[0].values.reshape(1,1))
```

```python
lr.predict(x_test.iloc[1].values.reshape(1,1))
```

```python
lr.predict(x_test.iloc[2].values.reshape(1,1))
```

# again plot the x train data

```python
plt.scatter(df['cgpa'],df['package'])
plt.plot(x_train,lr.predict(x_train),color='red')
plt.xlabel('CGPA')
plt.ylabel('pakage(in lpa)')
```

# find the value of m and b

```python
m = lr.coef_
```

```python
b = lr.intercept_
```

```python
m
```

```python
b
```

# calculate the slope line

```python
y = m*8.58+b
```
